## Exercise 1: Create a map showing the median of the a59j-regional-006 model for EEA ozone stations in Spain

```python
provi.filter("country", keep="Spain")
```

```python
provi.plot('map-Median', networkspeci='EEA|sconco3', labela='a59j-regional-006')
```

## Exercise 2: Create a map showing the RMSE of the rural nitrogen dioxide stations using the a59j-regional-006 model in Germany

```python
provi.reset()
```

```python
provi.filter("country", keep="Germany")
```

```python
provi.filter("area_classification", keep="rural")
```

```python
provi.plot('map-RMSE', labela='observations', labelb='a59j-regional-006', networkspeci='EEA|sconcno2')
```
