Get all the possible absolute paths to the nc files directories by filtering the domain, species, resolution, etc.

```python
def load_paths(pkl_file):
    with open(pkl_file, 'rb') as file:
        posibles_paths = pickle.load(file)
    return posibles_paths
```

```python
def search_paths(pkl_file):
    posibles_paths = []

    for experiment_dict in interp_experiments.values():
        print(experiment_dict)
        for exp in experiment_dict["models"]:
            for path in experiment_dict["paths"]:
                for dom in domain:
                    full_path = os.path.join(path,exp,dom)
                    isdir = os.path.isdir(full_path)
                    if not isdir:
                        continue
                        # if full_path.startswith("/gpfs"):
                        #     print(full_path)
                        # experiments.append(f"{exp}-{dom}")
                    for res in resolutions:
                        for spec in species:
                            full_path = os.path.join(path,exp,dom,res,spec)
                            isdir = os.path.isdir(full_path)
                            if isdir:
                                posibles_paths.append(full_path)

    with open(pkl_file, 'wb') as file:
        pickle.dump(posibles_paths, file)
    
    return posibles_paths
```

```python
pkl_file = 'posibles_paths.pkl'

if os.path.exists(pkl_file):
    posibles_paths = load_paths(pkl_file)
else:
    posibles_paths = search_paths(pkl_file)
```

Get the different nc files and analyse the format

```python
nc_files_dict = {}
for path in posibles_paths:
    for file in os.listdir(path):
        # filter the files that do not start with the species
        species = path.split("/")[-1]
        if file.startswith(species):
            nc_files_dict[file] = path
```

# Possible formats
Two main groups: 
1. species-ensopt_, 
    1. species-ensopt_date.nc E.g. loaddu-000_2023052013.nc (common)
    2. species-ensopt_date_ending.nc E.g. loaddu-000_2023052012_horizontal.nc Possible endings: {'diff.nc', 'horizontal.nc', 'an.nc'} (common)
2. species_
    1. species_date.nc E.g. loaddu_2023052013.nc (common)
    2. species_ensopt_date.nc E.g. loaddu_006_2023052013.nc (rare, error i guess)
    3. species_date_av_an.nc E.g. loaddu_2023052013_av_an.nc (rare, all endings are av_an )


species-ensopt_date.nc

```python
fun = lambda file :(file.count("-"),file.count("_")) == (1,1)
nc_files = list(filter(fun, nc_files_dict.keys()))
nc_files
```

```python
fun = lambda file :(file.count("-"),file.count("_")) == (1,2)
nc_files = list(filter(fun, nc_files_dict.keys()))
nc_files
```
